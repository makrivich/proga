```python
import pandas as pd
import hashlib

df_k = pd.read_csv('key.csv', low_memory=False)
df_m = pd.read_csv('message.csv')
df_c = pd.read_csv('codebook.csv')
N = len(df_c)
relays = df_k['relay'].unique()
```

```python
off = [r for r in relays if df_k[df_k['relay'] == r]['shift'].dropna().empty][0]

one_num_relay, one_num_val = None, None
cycle_relay, cycle_count = None, None

for r in relays:
    s = pd.to_numeric(df_k[df_k['relay'] == r]['shift'], errors='coerce').dropna()
    u = s.unique()
    if len(u) == 1:
        one_num_relay, one_num_val = r, int(u[0])
    elif 1 < len(u) <= 5:
        cycle_relay, cycle_count = r, len(u)
```

```python
garbage_relay, garbage_count = None, 0
for r in relays:
    raw = df_k[df_k['relay'] == r]['shift']
    cnt = raw.notna().sum() - pd.to_numeric(raw, errors='coerce').notna().sum()
    if cnt > 0:
        garbage_relay, garbage_count = r, cnt
```

```python
piv = df_k.copy()
piv['shift'] = pd.to_numeric(piv['shift'], errors='coerce')
tbl = piv.pivot(index='pos', columns='relay', values='shift')

m_pre = tbl.mode(axis=1)[0]
cand = [r for r in relays if r not in [off, one_num_relay, cycle_relay, garbage_relay]]
sneaky = [r for r in cand if (tbl[r] == m_pre).mean() < 0.5]
good = [r for r in cand if (tbl[r] == m_pre).mean() >= 0.5]
```

```python
key = [int(tbl[good].loc[p].mode().iloc[0]) for p in range(len(df_m))]

c2ch = dict(zip(df_c['code'], df_c['char']))
msg = df_m.sort_values('pos').reset_index(drop=True)
txt = "".join(c2ch[(r['code'] - key[i]) % N] for i, r in msg.iterrows())

pad_code = txt[-15:]
sum_key = sum(key)
sneaky_1, sneaky_2 = sneaky[0], sneaky[1]
```

```python
print(off)
print(one_num_relay)
print(one_num_val)
print(cycle_relay)
print(cycle_count)
print(garbage_relay)
print(garbage_count)
print(sneaky_1)
print(sneaky_2)
print(sum_key)
print(pad_code)
print(txt)
```
